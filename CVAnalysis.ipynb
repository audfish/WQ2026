```python
# importing needed packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#read in Central Valley csvs from external drive
path1 = '/Volumes/disk1/Central Valley Fog/4205333.csv'
path2 = '/Volumes/disk1/Central Valley Fog/4206426.csv'
path3 = '/Volumes/disk1/Central Valley Fog/4206427.csv'

df1 = pd.read_csv(path1)
df2 = pd.read_csv(path2)
df3 = pd.read_csv(path3)
```

```python
# print all column names
for col in df1.columns:
    print(col)
```

```python
df_list = [df1, df2, df3] 

CV_data = pd.concat(df_list, ignore_index = True)

print(CV_data.head(5))
```

```python
CV_fog = CV_data["MonthlyNumberDaysWithHeavyFog"]
CV_fog_rows = CV_fog[CV_fog.notna()]
print(CV_fog_rows)
```

```python
#make DF with dates and station names
CV_date_fog_columns = ["STATION", "DATE", "MonthlyNumberDaysWithHeavyFog"]

CV_date_fog_data = CV_data[CV_date_fog_columns]
print(CV_date_fog_data)
```

```python
CV_fog_rows = CV_date_fog_data[CV_date_fog_data.notna().all(axis=1)]
print(CV_fog_rows)
```

```python
CV_fog_rows["DATE"] = pd.to_datetime(CV_fog_rows["DATE"])
```

```python
CV_fog_rows["year"] = CV_fog_rows["DATE"].dt.year
CV_fog_rows["month"] = CV_fog_rows["DATE"].dt.month
print(CV_fog_rows.head(5))
```

```python
station_total_fog_days = CV_fog_rows.groupby(['STATION','year'])['month'].sum().rename('station_fog_days').reset_index()
CVtotal_fog_rows = station_total_fog_days.dropna()
print(station_total_fog_days)
print(CVtotal_fog_rows)
```

```python
unique_names = CVtotal_fog_rows['STATION'].unique()
names_df = pd.DataFrame(unique_names)
print(unique_names)
print(names_df)
```

```python
# Create subplots dynamically
fig, axs = plt.subplots(1, len(unique_names),  sharey=True)

# Iterate through unique values and plot on respective axes
for i in unique_names:
    CVtotal_fog_rows.loc[(CVtotal_fog_rows['STATION'] == i)].plot(x='year',y="station_fog_days", figsize=(15,4), kind = 'bar')
    plt.title(i, fontsize=18) #Labeling title
    plt.xlabel('years', fontsize=12) #Labeling x-axis 
    plt.ylabel('station fog days', fontsize=12) #Labeling y-axis 

plt.show()
```

```python
unique_station_total_fog_days = CVtotal_fog_rows.groupby(['STATION'])['year'].sum().rename('station_fog_days_total').reset_index()
print(unique_station_total_fog_days)
```

```python
#some sort of plot
fig1, ax1 = plt.subplots(figsize=(35,5))

br1 = ax1.bar(
    unique_station_total_fog_days["STATION"],
    unique_station_total_fog_days["station_fog_days_total"],
    alpha=0.75
)
# Set customizations
ax1.set_xlabel("Station")
ax1.set_ylabel("# days with heavy fog")
ax1.set_title("Central Valley station counts of number of days with heavy fog")


#Gives which stations have the most fog measurements to date. 
```
