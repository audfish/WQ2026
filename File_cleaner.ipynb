```python
#for each file in the folder path, print the column names
for filename in os.listdir(path):
    if filename.endswith(".psv"):
        full_path = os.path.join(path, filename)
        df = pd.read_csv(full_path, nrows=0, sep='|', low_memory=False)
        print(f"{full_path}:")
        print(df.columns.tolist())
        print()
```

```python
cols = ['STATION', 'Station_name', 'DATE', 'Year', 'Month', 'Day', 'Hour', 'Minute', 'LATITUDE', 'LONGITUDE','temperature','dew_point_temperature','wind_direction','wind_speed','precipitation', 'visibility']

dfs = {}
for filename in os.listdir(path):
    if filename.endswith(".psv"):
        full_path = os.path.join(path, filename)
        dfs[filename] = pd.read_csv(full_path, sep='|', usecols=cols, low_memory=False)
        print(f"{filename}: {dfs[filename].shape}")
```

```python
# concatenate the dataframes into one
all_data = pd.concat(dfs.values(), ignore_index=True)
print(f"Combined dataframe shape: {all_data.shape}")
all_data.head()
all_data.info()
```

```python
#convert date to datetime
all_data['DATE'] = pd.to_datetime(all_data['DATE'], errors='coerce')
all_data.info()
```

```python
# Rename columns
all_data.rename(columns={
    'STATION': 'STATION_ID',
    'Station_name': 'STATION',
    'Year': 'year',
    'Month': 'month',
    'Day': 'day',
    'Hour': 'hour',
    'Minute': 'minute',
    'temperature': 'HourlyDryBulbTemperature',
    'dew_point_temperature': 'HourlyDewPointTemperature',
    'wind_direction': 'HourlyWindDirection',
    'wind_speed': 'HourlyWindSpeed',
    'precipitation': 'HourlyPrecipitation',
    'visibility': 'HourlyVisibility'
}, inplace=True)
```

```python
stations_to_keep = ['BAKERSFIELD AP', 'BEALE AFB', 'CASTLE AFB', 'FRESNO YOSEMITE INTL', 'LEMOORE REEVES NAS', 'REDDING AP', 'SACRAMENTO AP ASOS', 'SACRAMENTO MCCLELLAN AFB', 'SACRAMENTO MATHER AFB', 'SACRAMENTO METRO AP', 'STOCKTON AP', 'TRAVIS FLD AFB']

all_data = all_data[all_data['STATION'].isin(stations_to_keep)].reset_index(drop=True)

print(f"Filtered dataframe shape: {all_data.shape}")
print(all_data['STATION'].unique())
```

```python
all_data.head()
```

```python
# Convert DATE from UTC to UTC-7
# does this add or subtract time from the original DATE? It subtracts 7 hours to convert from UTC to UTC-7.
all_data['DATE'] = all_data['DATE'] - pd.Timedelta(hours=7)

# Update hour, day, month, year columns to match the adjusted DATE
all_data['hour'] = all_data['DATE'].dt.hour
all_data['day'] = all_data['DATE'].dt.day
all_data['month'] = all_data['DATE'].dt.month
all_data['year'] = all_data['DATE'].dt.year

all_data.head()
```

```python
#make sure values are numeric
numeric_cols = ['HourlyDryBulbTemperature','HourlyDewPointTemperature','HourlyWindDirection','HourlyWindSpeed','HourlyPrecipitation', 'HourlyVisibility']
all_data[numeric_cols] = all_data[numeric_cols].apply(pd.to_numeric, errors='coerce')
all_data.info()
```

```python
# Fill NaN values for HourlyPrecipitation as 0
all_data['HourlyPrecipitation'] = all_data['HourlyPrecipitation'].fillna(0)

# show to check
all_data['HourlyPrecipitation'].head()
```

```python
#cap HourlyVisibility values at 16
all_data['HourlyVisibility'] = all_data['HourlyVisibility'].clip(upper=16)
```

```python
# If HourlyWindSpeed is equal to zero, set HourlyWindDirection to NaN
all_data.loc[all_data['HourlyWindSpeed'] == 0, 'HourlyWindDirection'] = np.nan

#print the rows where HourlyWindSpeed is zero to verify the change
print(all_data.loc[all_data['HourlyWindSpeed'] == 0, ['HourlyWindSpeed', 'HourlyWindDirection']])
all_data.shape
```

```python
# Print negative hourlywindspeed values
print(all_data.loc[all_data['HourlyWindSpeed'] < 0, ['HourlyWindSpeed', 'HourlyWindDirection']])
all_data.shape
```

```python
# If HourlyWindSpeed is equal to NaN, set HourlyWindDirection to NaN
all_data.loc[all_data['HourlyWindSpeed'].isna(), 'HourlyWindDirection'] = np.nan

#print the rows where HourlyWindSpeed is NaN to verify the change
print(all_data.loc[all_data['HourlyWindSpeed'].isna(), ['HourlyWindSpeed', 'HourlyWindDirection']])
all_data.shape
```
