```python
#for each file in the folder path, print the column names
for filename in os.listdir(path):
    if filename.endswith(".psv"):
        full_path = os.path.join(path, filename)
        df = pd.read_csv(full_path, nrows=0, sep='|', low_memory=False)
        print(f"{full_path}:")
        print(df.columns.tolist())
        print()
```

```python
cols = ['STATION', 'Station_name', 'DATE', 'Year', 'Month', 'Day', 'Hour', 'Minute', 'LATITUDE', 'LONGITUDE','temperature','dew_point_temperature','wind_direction','wind_speed','precipitation', 'visibility']

dfs = {}
for filename in os.listdir(path):
    if filename.endswith(".psv"):
        full_path = os.path.join(path, filename)
        dfs[filename] = pd.read_csv(full_path, sep='|', usecols=cols, low_memory=False)
        print(f"{filename}: {dfs[filename].shape}")
```

```python
# concatenate the dataframes into one
all_data = pd.concat(dfs.values(), ignore_index=True)
print(f"Combined dataframe shape: {all_data.shape}")
all_data.head()
all_data.info()
```

```python
# Rename columns
all_data.rename(columns={
    'STATION': 'STATION_ID',
    'Station_name': 'STATION',
    'Year': 'year',
    'Month': 'month',
    'Day': 'day',
    'Hour': 'hour',
    'Minute': 'minute',
    'temperature': 'HourlyDryBulbTemperature',
    'dew_point_temperature': 'HourlyDewPointTemperature',
    'wind_direction': 'HourlyWindDirection',
    'wind_speed': 'HourlyWindSpeed',
    'precipitation': 'HourlyPrecipitation',
    'visibility': 'HourlyVisibility'
}, inplace=True)
```

```python
#make sure values are numeric
numeric_cols = ['HourlyDryBulbTemperature','HourlyDewPointTemperature','HourlyWindDirection','HourlyWindSpeed','HourlyPrecipitation', 'HourlyVisibility']
all_data[numeric_cols] = all_data[numeric_cols].apply(pd.to_numeric, errors='coerce')
all_data.info()
```

```python
#convert date to datetime
all_data['DATE'] = pd.to_datetime(all_data['DATE'], errors='coerce')
all_data.info()
```

```python
# Fill NaN values for HourlyPrecipitation as 0
all_data['HourlyPrecipitation'] = all_data['HourlyPrecipitation'].fillna(0)

# show to check
all_data['HourlyPrecipitation'].head()
```

```python
#cap HourlyVisibility values at 16
all_data['HourlyVisibility'] = all_data['HourlyVisibility'].clip(upper=16)
```

```python
# Save as CSV
all_data.to_csv('cleaned_data_Oct.csv', index=False)
```
