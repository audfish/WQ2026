```python
# create dataframes with only the columns we need
# units: HourlyVisibility: km to thousandths, HourlyPrecipitation: mm to tenths

df1 = pd.read_csv(path1, usecols=['STATION', 'DATE', 'REPORT_TYPE', 'LATITUDE', 'LONGITUDE', 'HourlyVisibility','HourlyPrecipitation','HourlyDryBulbTemperature','HourlyDewPointTemperature', 'HourlyWindDirection', 'HourlyWindSpeed', 'MonthlyNumberDaysWithHeavyFog'],low_memory=False)
df2 = pd.read_csv(path2, usecols=['STATION', 'DATE', 'REPORT_TYPE', 'LATITUDE', 'LONGITUDE', 'HourlyVisibility','HourlyPrecipitation','HourlyDryBulbTemperature','HourlyDewPointTemperature', 'HourlyWindDirection', 'HourlyWindSpeed', 'MonthlyNumberDaysWithHeavyFog'],low_memory=False)
# df1 = pd.read_csv(path1, usecols=['STATION', 'DATE', 'LATITUDE', 'LONGITUDE', 'HourlyVisibility','HourlyPrecipitation','MonthlyNumberDaysWithHeavyFog'],low_memory=False)
# df2 = pd.read_csv(path2, usecols=['STATION', 'DATE', 'LATITUDE', 'LONGITUDE', 'HourlyVisibility','HourlyPrecipitation','MonthlyNumberDaysWithHeavyFog'],low_memory=False)
#df3 = pd.read_csv(path3, usecols=['STATION', 'DATE', 'LATITUDE', 'LONGITUDE', 'DailySustainedWindDirection', 'HourlyVisibility','HourlyPrecipitation','MonthlyNumberDaysWithHeavyFog'],low_memory=False)
```

```python
#df_list = [df1, df2, df3] 
df_list = [df1, df2] 

# concatenate the dataframes into one
CV_data = pd.concat(df_list, ignore_index = True)

print(CV_data.head(15))
```

```python
# convert HourlyVisibility, HourlyDewPointTemperature, and HourlyDryBulbTemperature to numeric, coercing errors to NaN
CV_data["HourlyVisibility"] = pd.to_numeric(CV_data["HourlyVisibility"], errors="coerce")
CV_data['HourlyDewPointTemperature'] = pd.to_numeric(CV_data['HourlyDewPointTemperature'], errors='coerce')
CV_data['HourlyDryBulbTemperature'] = pd.to_numeric(CV_data['HourlyDryBulbTemperature'], errors='coerce')
CV_data['HourlyWindSpeed'] = pd.to_numeric(CV_data['HourlyWindSpeed'], errors='coerce')
CV_data['HourlyWindDirection'] = pd.to_numeric(CV_data['HourlyWindDirection'], errors='coerce')
CV_data['HourlyPrecipitation'] = pd.to_numeric(CV_data['HourlyPrecipitation'], errors='coerce')

# vis_nonpos = CV_data_filtered[CV_data_filtered["HourlyVisibility"] <= 0]
# print(f"Rows with HourlyVisibility <= 0: {len(vis_nonpos)}")
# vis_nonpos.head(100)

CV_data.head(15)
```

```python
# fill in missing values for HourlyWindDirection and HourlyWindSpeed with 0
# CV_data["HourlyVisibility"] = CV_data["HourlyVisibility"].fillna(16.000)
CV_data["HourlyWindDirection"] = CV_data["HourlyWindDirection"].fillna(0)
CV_data["HourlyWindSpeed"] = CV_data["HourlyWindSpeed"].fillna(0)
CV_data["HourlyPrecipitation"] = CV_data["HourlyPrecipitation"].fillna(0)

CV_data.head(15)
```

```python
# Cap visibilities over 16.000 km to 16.000 km in CV_rows
CV_data['HourlyVisibility'] = CV_data['HourlyVisibility'].clip(upper=16.000)

print("CV_data after capping visibility at 16.000 km:")
print(f"Max visibility: {CV_data['HourlyVisibility'].max()}")
print(f"Min visibility: {CV_data['HourlyVisibility'].min()}")
```

```python
# convert date column to datetime format
CV_data["DATE"] = pd.to_datetime(CV_data["DATE"])
```

```python
# extract year and month from date column and add as new columns
CV_data["year"] = CV_data["DATE"].dt.year
CV_data["month"] = CV_data["DATE"].dt.month
CV_data["day"] = CV_data["DATE"].dt.day
CV_data["hour"] = CV_data["DATE"].dt.hour
CV_data["minute"] = CV_data["DATE"].dt.minute
print(CV_data.head(15))
```

```python
# read in station names CSV in order to change from station codes to station names
path = 'E:/Central Valley Fog/station_names.csv'
# path = '/Volumes/disk1/Central Valley Fog/station_names.csv'

station_names = pd.read_csv(path, low_memory = False)
# station_names = pd.read_csv(path, low_memory = False)


print(station_names)
```

```python
# change station ID to station name in CV_rows
station_id_to_name = station_names.set_index('STATION_ID')['STATION']

CV_data = CV_data.rename(columns={'STATION': 'STATION_ID'})
CV_data['STATION'] = CV_data['STATION_ID'].map(station_id_to_name).fillna(CV_data['STATION_ID'])

# check:
print(CV_data[['STATION_ID', 'STATION']].head())
```
