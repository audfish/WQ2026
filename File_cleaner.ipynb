```python
# create dataframes with only the columns we need
# units: HourlyVisibility: km to thousandths, HourlyPrecipitation: mm to tenths
df1 = pd.read_csv(path1, usecols=['STATION', 'DATE', 'LATITUDE', 'LONGITUDE', 'HourlyVisibility','HourlyPrecipitation','MonthlyNumberDaysWithHeavyFog'],low_memory=False)
df2 = pd.read_csv(path2, usecols=['STATION', 'DATE', 'LATITUDE', 'LONGITUDE', 'HourlyVisibility','HourlyPrecipitation','MonthlyNumberDaysWithHeavyFog'],low_memory=False)
#df3 = pd.read_csv(path3, usecols=['STATION', 'DATE', 'LATITUDE', 'LONGITUDE', 'HourlyVisibility','HourlyPrecipitation','MonthlyNumberDaysWithHeavyFog'],low_memory=False)
```

```python
#df_list = [df1, df2, df3] 
df_list = [df1, df2] 

CV_data = pd.concat(df_list, ignore_index = True)

print(CV_data.head(15))
```

```python
# filter for rows that have hourly visibility data
CV_rows = CV_data[CV_data['HourlyVisibility'].notna()]
print(CV_rows)
```

```python
#fill in missing values for HourlyPrecipitation with 0
CV_rows["HourlyPrecipitation"] = CV_rows["HourlyPrecipitation"].fillna(0)
```

```python
# convert date column to datetime format
CV_rows["DATE"] = pd.to_datetime(CV_rows["DATE"])
```

```python
# extract year and month from date column and add as new columns
CV_rows["year"] = CV_rows["DATE"].dt.year
CV_rows["month"] = CV_rows["DATE"].dt.month
CV_rows["day"] = CV_rows["DATE"].dt.day
CV_rows["hour"] = CV_rows["DATE"].dt.hour
CV_rows["minute"] = CV_rows["DATE"].dt.minute
print(CV_rows.head(15))
```

```python
# convert hourly precipitation and visibility to numeric values
CV_rows["HourlyPrecipitation"] = pd.to_numeric(CV_rows["HourlyPrecipitation"], errors='coerce')
CV_rows["HourlyVisibility"] = pd.to_numeric(CV_rows["HourlyVisibility"], errors='coerce')
```

```python
# drop MonthlyNumberDaysWithHeavyFog column from CV_rows
CV_rows = CV_rows.drop(columns=['MonthlyNumberDaysWithHeavyFog'])

CV_rows.head()
```

```python
# save cleaned dataframe
CV_rows.to_csv('CV_rows_cleaned.csv', index=False)
```
